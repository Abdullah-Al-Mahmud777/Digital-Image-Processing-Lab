```python
#Experiment Name: Color space Conversion
import PIL.Image as Image
import matplotlib.pyplot as plt
img1=Image.open("image.jpg")
plt.imshow(img1)
plt.title("Picture 1")
plt.show()
```

```python
# Image Property Analysis
Width, Height = img1.size
print("Width:", Width)
print("Height:", Height)
print(img1.mode)
```

```python
#Displaying Multiple Images Using Subplots
import matplotlib.pyplot as plt

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Display first image
axes[0].imshow(img1)
axes[0].set_title("Original Image")

# Display second image (if available)
img2 = img1.convert("L")  # Convert to grayscale for demonstration
axes[1].imshow(img2, cmap='gray')
axes[1].set_title("Grayscale Image")

plt.tight_layout()
plt.show()
```

```python
 #Conversion from RGB to Grayscale Color Space
img_gray = img1.convert("L")
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale Image")
plt.show()

```

```python
#image 2 Conversion from RGB to Grayscale Color Space
img2_gray = img2.convert("L")
plt.imshow(img2_gray, cmap='gray')
plt.title("Grayscale Image 2")
plt.show()
```
