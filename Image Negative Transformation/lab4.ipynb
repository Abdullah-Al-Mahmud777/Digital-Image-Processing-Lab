```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# a) Read a grayscale image

img = cv2.imread('image.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: Image not found!")
else:
    # b) Extract a 5 x 5 pixel region from the top-left corner
    roi_original = img[0:5, 0:5]
    # c) Print the intensity values of that 5 x 5 region
    print("Original Image 5x5 Region Intensity Values:")
    print(roi_original)
```

```python
# d) Apply image negative transformation

img_negative = 255 - img

# e) Print the corresponding 5 x 5 region from the negative image

roi_negative = img_negative[0:5, 0:5]

print("\nNegative Image 5x5 Region Intensity Values:")
print(roi_negative)

```

```python
# f) Display both images
plt.figure(figsize=(6, 6))

# Original Grayscale Image
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

# Negative Image
plt.subplot(1, 2, 2)
plt.imshow(img_negative, cmap='gray')
plt.title('Negative Transformed Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
