```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# a) Read a grayscale image
img = cv2.imread('kk.png', cv2.IMREAD_GRAYSCALE)

# b) Use replicate padding (edge padding) to handle boundary pixels
# OpenCV's filter2D and medianBlur use replicate padding by default

# c) Apply a 3x3 averaging filter to smooth the image
kernel = np.ones((3, 3), np.float32) / 9
avg_filtered = cv2.filter2D(img, -1, kernel, borderType=cv2.BORDER_REPLICATE)

# d) Apply a 3x3 median filter to smooth the image
median_filtered = cv2.medianBlur(img, 3)

# e) Display the Original image, Averaging filter output, and Median filter output
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title('Averaging Filter')
plt.imshow(avg_filtered, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('Median Filter')
plt.imshow(median_filtered, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
